```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_diabetes
```

```python
dibates = load_diabetes()
dibates_df = pd.DataFrame(dibates.data,columns=dibates.feature_names)
```

```python
dibates_df['target'] = dibates.target
print(dibates_df.head())
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(dibates_df.corr(),annot=True)
```

```python
features = dibates_df.columns[0:11]
target = dibates_df.columns[-1]

X = dibates_df[features].values
y = dibates_df[target].values

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=17)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
model = LinearRegression()
model
```

```python
model.fit(X_train,y_train)
```

```python
prediction = model.predict(X_test)
```

```python
actual = y_test
```

```python
train_score = model.score(X_train,y_train)
test_score = model.score(X_test,y_test)
```

```python
print(f'The Train Score '+ str(train_score))
print(f'The Test Score ' + str(test_score))
```
